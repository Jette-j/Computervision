```python
def snake(img, contour, **kwargs):
    # parameter initialization
    k = contour.shape[1]
    alpha = kwargs['alpha'] * np.ones(k)
    beta = kwargs['beta'] * np.ones(k)
    gamma = kwargs['gamma'] * np.ones(k)

    # TODO: compute gradient image
    gradient_x, gradient_y = gradient(img)
    gradient_magnitude = np.sqrt(gradient_x**2 + gradient_y**2)
    # Normalize gradient magnitude to [0, 1]
    g_min = gradient_magnitude.min()
    g_max = gradient_magnitude.max()
    norm_image_gradient = (gradient_magnitude - g_min) / (g_max - g_min + 1e-8)

    k_updated = k
    iterations = 0
    # iterate until snake stops changing
    while k_updated > k * kwargs['contour_fraction']:
        iterations = iterations + 1
        k_updated = 0

        # update all contour points
        for i in range(k):
            p, updated = greedy_minimization(
                norm_image_gradient, contour, i, alpha, beta, gamma, kwargs['neighborhood_size'])
            if p is not None:
                contour[:, i] = p
            # eliminate corners if required
            if iterations > kwargs['begin_corner_elim']:
                beta = corner_elimination(contour, beta, kwargs['K_threshold'])

            # keep track of number of updated points
            if updated:
                k_updated = k_updated + 1

        # display current state of snake
        print('num_updates: {}'.format(k_updated))
        plt.imshow(norm_image_gradient)
        plt.plot(contour[1, :], contour[0, :], 'r.-')
        plt.show()
```

```python
def greedy_minimization(norm_image_gradient, contour, i, alpha, beta, gamma, neighborhood_size):
    m, n = norm_image_gradient.shape
    k = contour.shape[1]
    w = (kwargs['neighborhood_size'] - 1.) // 2.
    d = average_distance(contour)
    updated = False

    # get previous, current, and next point in snake
    p_iminus1 = contour[:, wrap_index(i-1, k)]
    p_i = contour[:, i]
    p_iplus1 = contour[:, wrap_index(i+1, k)]

    # compute individual terms of energy functional
    p = []
    E_cont = []
    E_curv = []
    E_imag = []

    # For all points in a local neighborhood, compute the energy terms
    for p_i_x in np.arange(max(0, p_i[0] - w), min(m-1, p_i[0]+w)+1):
        for p_i_y in np.arange(max(0, p_i[1]-w), min(n-1, p_i[1]+w)+1):
            pass
            # Get the candidate point vector
            p_candidate = np.array([p_i_x, p_i_y])

            # E_cont: penalizes deviation from uniform spacing
            # Uses (0.5 * (||p_i - p_{i-1}|| - ||p_i - p_{i+1}||))^2
            dist_prev = norm(p_candidate - p_iminus1)
            dist_next = norm(p_candidate - p_iplus1)
            e_cont = (0.5 * (dist_prev - dist_next)) ** 2

            # E_curv: penalizes curvature (second finite difference)
            e_curv = norm(p_iminus1 - 2 * p_candidate + p_iplus1) ** 2

            # E_imag: negative gradient magnitude (snake attracted to edges)
            e_imag = -norm_image_gradient[int(p_i_x), int(p_i_y)]

            p.append(p_candidate)
            E_cont.append(e_cont)
            E_curv.append(e_curv)
            E_imag.append(e_imag)

    if not p:
        return None, False

    # TODO: normalize individual terms of energy functional
    def normalize(E):
        E = np.array(E, dtype=float)
        e_min, e_max = E.min(), E.max()
        if e_max - e_min < 1e-8:
            return np.zeros_like(E)
        return (E - e_min) / (e_max - e_min)

    E_cont_norm = normalize(E_cont)
    E_curv_norm = normalize(E_curv)
    E_imag_norm = normalize(E_imag)

    # Weighted sum of energy terms
    E_total = (alpha[i] * E_cont_norm
             + beta[i]  * E_curv_norm
             + gamma[i] * E_imag_norm)

    # compute energy functional and determine minimum
    min_idx = np.argmin(E_total)
    new_p_i = p[min_idx]
    # determine if the current point of snake was moved
    if (p_i - new_p_i).sum() != 0:
        updated = True

    return new_p_i, updated
```

```python
def gradient(image):
    v = .5
    k_y = np.zeros((3, 3))
    k_y[0, 1] = -v
    k_y[2, 1] = +v

    k_x = np.zeros((3, 3))
    k_x[1, 0] = -v
    k_x[1, 2] = +v

    image = np.copy(image).astype('float32')

    gradient_x = cv2.filter2D(image, -1, k_x)
    gradient_y = cv2.filter2D(image, -1, k_y)

    return gradient_x, gradient_y
```

```python
def load_image_gray_normalized(image_path):
    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    return image.astype('float32')/255.
```

```python
def wrap_index(index, k):
    if index < 0:
        index = k - 1
    elif index > k-1:
        index = 0
    return index
```

```python
def average_distance(contour):
    avg_dist = 0
    n = contour.shape[1]

    for i in range(n-1):
        avg_dist = avg_dist+norm(contour[:, i] - contour[:, i+1])

    avg_dist = avg_dist+norm(contour[:, -1] - contour[:, 0])
    avg_dist = avg_dist / n

    return avg_dist
```

```python
def corner_elimination(contour, beta, K_threshold):
    n = contour.shape[1]
    K = []
    # compute curvature of snake at each point
    for i in range(n):
        # get previous, current, and next point in snake
        p_iminus1 = contour[:, wrap_index(i-1, n)]
        p_i = contour[:, i]
        p_iplus1 = contour[:, wrap_index(i+1, n)]
        K.append(((p_iminus1 - 2.*p_i + p_iplus1)**2.).sum())

    # determine local maxima and set beta to zero there
    for i in range(n):
        prev_is_lower = K[wrap_index(i-1, n)] < K[i]
        next_is_lower = K[i] > K[wrap_index(i+1, n)]
        thres_exceeded = K[i] > K_threshold
        if prev_is_lower and next_is_lower and thres_exceeded:
            beta[i] = 0

    return beta
```

The lecture formulation penalizes the deviation of the distance from point i to its predecessor from the global average spacing d̄. This actively drives points toward uniform distribution across the whole contour, but requires computing and re-using d̄ each iteration — and can be destabilizing when the snake undergoes large deformations.
The exercise formulation instead measures the difference between the preceding gap and the following gap around point i. It is zero whenever the point sits exactly halfway between its two neighbours, regardless of what the absolute spacing is. This is a purely local criterion: it does not compare against a global reference and therefore does not pull the snake toward any particular overall scale. As a result it is somewhat weaker at enforcing truly uniform spacing globally, but it is more stable and self-contained — no global statistic needs to be maintained — and it still effectively prevents clustering of points in one region while others are sparsely covered.
In practice, both terms serve the same qualitative purpose (spacing regularisation), but the lecture version is more aggressive about matching a target inter-point distance, while the exercise version is a softer, purely relative balance condition.
